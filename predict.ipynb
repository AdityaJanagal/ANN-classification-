```python
from tensorflow.keras.models import load_model
import pickle
import pandas as pd 
import numpy as np

model = load_model('model.h5')

with open('label_encoder_gender.pkl','rb') as file:
    label_encoder_gender= pickle.load(file)
    
with open('One_hot_encoder_geo.pkl','rb') as file:
    One_hot_encoder_geo= pickle.load(file)

with open('scaler.pkl','rb') as file:
    scaler= pickle.load(file)
```

```python
input_data={
    "RowNumber": 1250,
    "CustomerId": 15738492,
    "Surname": "Smith",
    "CreditScore": 742,
    "Geography": "Germany",
    "Gender": "Male",
    "Age": 35,
    "Tenure": 6,
    "Balance": 125430.75,
    "NumOfProducts": 2,
    "HasCrCard": 1,
    "IsActiveMember": 1,
    "EstimatedSalary": 87342.56,
    "Exited": 0
}
```

```python
input_data_df = pd.DataFrame([input_data])
y = input_data_df['Exited']
input_data_df = input_data_df.drop(columns=[ 'RowNumber','CustomerId','Surname','Exited'])
```

```python
input_data_df['Geography']
```

```python
encoder_geo = One_hot_encoder_geo.transform(input_data_df[['Geography']].astype(str))
encoded_df = pd.DataFrame(encoder_geo,columns=One_hot_encoder_geo.get_feature_names_out(['Geography']))
encoded_df
```

```python
label_gender = label_encoder_gender.transform(input_data_df[['Gender']])
label_gender_df['Gender'] = pd.DataFrame(label_gender)
label_gender_df
```

```python
input_data_df=input_data_df.drop(columns = ['Geography','Gender'])
input_data_df = pd.concat(
    [
    encoded_df,input_data_df,label_gender_df],axis=1
)
input_data_df
```

```python
input_data_df = input_data_df.drop(columns=[0])
```

```python
input_data_df = input_data_df[scaler.feature_names_in_]
input_data_df.columns.to_list()
```

```python
scaled_input = scaler.transform(input_data_df)
```

```python
prediction=model.predict(scaled_input)
prediction
```

```python
predict_prob = prediction[0][0]
predict_prob
```

```python
if predict_prob > 0.5:
    print("The customer is likely churn .")
else:
    print("The customer is not likely churn .")
```
